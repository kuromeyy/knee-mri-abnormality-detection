```python
import os, pandas as pd

path = None
for root, dirs, files in os.walk("/kaggle/input"):
    dirs[:] = [d for d in dirs if d != "competitions"]
    if "_meta_v2.csv" in files:
        path = os.path.join(root, "_meta_v2.csv"); break
print("Meta:", path)

m = pd.read_csv(path).drop_duplicates(["StudyInstanceUID", "view"])
m = m[m["ok"] == 1]
print(f"Series terpakai: {len(m)} dari {m['StudyInstanceUID'].nunique()} studi\n")

print("=== Jumlah irisan asli per view ===")
print(m.groupby("view")["n_slices"].describe(percentiles=[.1, .25, .5, .75, .9]).round(1).to_string())

print("\n=== Proporsi series yang irisannya lebih dari ... ===")
for t in [16, 24, 32]:
    print(f"> {t:2d} irisan:", (m["n_slices"] > t).groupby(m["view"]).mean().round(2).to_dict())
```
